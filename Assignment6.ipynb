```python
import pandas as pd
import numpy as np
```

```python
from sklearn import preprocessing
```

```python
# 1. Create the dataset manually
data=pd.DataFrame({
    'age':[15,16,17,18,19,20],'studytime':[1,2,3,2,4,3],'failures':[0,1,0,2,1,0]
}).astype(float)
```

```python
# 2. Min-Max Normalization
# Convert the values into the range[0,1]
minmax_scaler=preprocessing.MinMaxScaler()
minmax_scaled=minmax_scaler.fit_transform(data)
```

```python
# 3. z-Score Normalization
# mean=0 and standard deviation=1
zscore_scaled=preprocessing.scale(data)
```

```python
# 4.Decimal Scaling
j=np.ceil(np.log10(data.abs().max()))
decimal_scaled=(data/10**j)
```

```python
print(data)
```

```python
print(" Min-Max NORMALIZATION   ")
print(pd.DataFrame(minmax_scaled,columns=data.columns))
```

```python
print("Z-SCORED NORMALIZATION")
print(pd.DataFrame(zscore_scaled,columns=data.columns))
```

```python
print("DECIMAL SCALING NORMALIZATION")
print(pd.DataFrame(decimal_scaled,columns=data.columns))
```
